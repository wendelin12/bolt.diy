```python
# 1. Notwendige GIS-Bibliotheken automatisch installieren
!pip install -q osmnx matplotlib geopandas shapely

import osmnx as ox
import matplotlib.pyplot as plt

# 2. Augsburg als Ziel definieren
place_name = "Augsburg, Germany"
print(f"Lade Geodaten für {place_name}...")

# 3. Straßennetz für Augsburg herunterladen
graph = ox.graph_from_place(place_name, network_type="drive")

print("Rendere minimalistische Architektur-Karte im Noir-Stil...")

# 4. Minimalistisches Design anwenden (Schwarzer Hintergrund, feine weiße Linien)
fig, ax = ox.plot_graph(
    graph,
    figsize=(12, 12),
    node_size=0,
    edge_color="#ffffff",
    edge_linewidth=0.6,
    bgcolor="#0b0b0b"
)

# 5. Direkt in Colab anzeigen lassen
plt.show()
```
